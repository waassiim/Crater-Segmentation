```python
import tensorflow as tf
from tensorflow.keras.layers import Input, Conv2D, AveragePooling2D, Dropout, UpSampling2D, Add
from tensorflow.keras.models import Model
from tensorflow.keras import backend as K

# --- 1. Define Custom Metrics ---

def dice_coef(y_true, y_pred):
    """
    Dice Coefficient (F1 Score)
    2 * (Intersection) / (Total Area)
    """
    smooth = 1e-6
    # Flatten the tensors
    y_true_f = K.flatten(y_true)
    y_pred_f = K.flatten(y_pred)
    
    intersection = K.sum(y_true_f * y_pred_f)
    return (2. * intersection + smooth) / (K.sum(y_true_f) + K.sum(y_pred_f) + smooth)

def iou_coef(y_true, y_pred):
    """
    Intersection over Union (Jaccard Index)
    (Intersection) / (Union)
    """
    smooth = 1e-6
    y_true_f = K.flatten(y_true)
    y_pred_f = K.flatten(y_pred)
    
    intersection = K.sum(y_true_f * y_pred_f)
    union = K.sum(y_true_f) + K.sum(y_pred_f) - intersection
    return (intersection + smooth) / (union + smooth)

def precision_m(y_true, y_pred):
    """Precision: True Positives / (True Positives + False Positives)"""
    smooth = 1e-6
    y_true_f = K.flatten(y_true)
    y_pred_f = K.flatten(y_pred)
    
    true_positives = K.sum(y_true_f * K.round(y_pred_f))
    predicted_positives = K.sum(K.round(y_pred_f))
    return (true_positives + smooth) / (predicted_positives + smooth)

def recall_m(y_true, y_pred):
    """Recall: True Positives / (True Positives + False Negatives)"""
    smooth = 1e-6
    y_true_f = K.flatten(y_true)
    y_pred_f = K.flatten(y_pred)
    
    true_positives = K.sum(y_true_f * K.round(y_pred_f))
    possible_positives = K.sum(y_true_f)
    return (true_positives + smooth) / (possible_positives + smooth)

# --- 2. The Model Architecture ---

def default_unet(): 
     z1 = Input(shape=(512,512,1))

     # Encoder path
     z2 = Conv2D(16, 3, padding='same', activation='relu')(z1)
     p2 = AveragePooling2D(pool_size=2)(z2)

     z3 = Conv2D(24, 3, padding='same', activation='relu')(p2)
     p3 = AveragePooling2D(pool_size=2)(z3)

     z4 = Conv2D(32, 3, padding='same', activation='relu')(p3)
     d4 = Dropout(0.2)(z4)
     p4 = AveragePooling2D(pool_size=2)(d4)

     z5 = Conv2D(48, 3, padding='same', activation='relu')(p4)
     d5 = Dropout(0.2)(z5)
     p5 = AveragePooling2D(pool_size=2)(d5)

     z6 = Conv2D(64, 3, padding='same', activation='relu')(p5)
     d6 = Dropout(0.2)(z6)
     p6 = AveragePooling2D(pool_size=2)(d6)

     z7 = Conv2D(96, 3, padding='same', activation='relu')(p6)
     d7 = Dropout(0.2)(z7)
     p7 = AveragePooling2D(pool_size=2)(d7)

     z8 = Conv2D(128, 3, padding='same', activation='relu')(p7)
     d8 = Dropout(0.2)(z8)
     p8 = AveragePooling2D(pool_size=2)(d8)

     z9 = Conv2D(128, 3, padding='same', activation='relu')(p8)
     d9 = Dropout(0.2)(z9)

     # Decoder path
     u9 = UpSampling2D(size=2)(d9)
     q9 = Conv2D(128, 3, padding='same', activation='relu')(u9)
     d9b = Dropout(0.2)(q9)
     a9 = Add()([d9b,z8])

     u8 = UpSampling2D(size=2)(a9)
     q8 = Conv2D(96, 3, padding='same', activation='relu')(u8)
     d8b = Dropout(0.2)(q8)
     a8 = Add()([d8b,z7])

     u7 = UpSampling2D(size=2)(a8)
     q7 = Conv2D(64, 3, padding='same', activation='relu')(u7)
     d7b = Dropout(0.2)(q7)
     a7 = Add()([d7b,z6])

     u6 = UpSampling2D(size=2)(a7)
     q6 = Conv2D(48, 3, padding='same', activation='relu')(u6)
     d6b = Dropout(0.2)(q6)
     a6 = Add()([d6b,z5])

     u5 = UpSampling2D(size=2)(a6)
     q5 = Conv2D(32, 3, padding='same', activation='relu')(u5)
     d5b = Dropout(0.2)(q5)
     a5 = Add()([d5b,z4])

     u4 = UpSampling2D(size=2)(a5)
     q4 = Conv2D(24, 3, padding='same', activation='relu')(u4)
     a4 = Add()([q4,z3])

     u3 = UpSampling2D(size=2)(a4)
     q3 = Conv2D(16, 3, padding='same', activation='relu')(u3)
     a3 = Add()([q3,z2])

     z_final = Conv2D(1, 3, padding='same', activation='sigmoid')(a3)

     model = Model(inputs = z1, outputs = z_final)
     
     # Compile with the new metrics
     model.compile(
         optimizer='adam', 
         loss='binary_crossentropy', 
         metrics=['accuracy', dice_coef, iou_coef, precision_m, recall_m]
     )
     
     return model
```

```python
import matplotlib.pyplot as plt

def plot_all_metrics(history):
    # 1. Define the metrics we want to plot
    # These names must match the function names used in model.compile
    metrics = [
        'loss', 
        'accuracy', 
        'dice_coef', 
        'iou_coef', 
        'precision_m', 
        'recall_m'
    ]
    
    # 2. Setup the plot grid (2 rows, 3 columns)
    fig, axes = plt.subplots(2, 3, figsize=(18, 10))
    axes = axes.flatten() # Flatten 2D array to 1D for easy iteration

    # 3. Loop through metrics and plot
    for i, metric in enumerate(metrics):
        ax = axes[i]
        
        # Check if the metric actually exists in the history object
        if metric in history.history:
            # Get training and validation values
            train_values = history.history[metric]
            val_values = history.history.get(f'val_{metric}') # Use .get() to be safe
            
            # Plotting
            epochs = range(1, len(train_values) + 1)
            ax.plot(epochs, train_values, 'b', label=f'Training {metric}')
            if val_values:
                ax.plot(epochs, val_values, 'r', label=f'Validation {metric}')
            
            # Styling
            ax.set_title(f'Training and Validation {metric.capitalize()}')
            ax.set_xlabel('Epochs')
            ax.set_ylabel(metric)
            ax.legend()
            ax.grid(True)
        else:
            # Hide empty subplots if a metric wasn't calculated
            ax.axis('off')

    plt.tight_layout()
    plt.show()

# --- HOW TO RUN IT ---
# Assuming your training history variable is named 'resultss'
plot_all_metrics(resultss)
```

```python
import matplotlib.pyplot as plt

# Plot utility
def plot_graphs(history, string):
  plt.plot(history.history[string])
  plt.plot(history.history['val_'+string])
  plt.xlabel("Epochs")
  plt.ylabel(string)
  plt.legend([string, 'val_'+string])
  plt.show()

# Plot the accuracy and results 
plot_graphs(resultss, "accuracy")
plot_graphs(resultss, "loss")
```
